```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))
if num1>num2:
    print(f"{num1} is large number")
elif num2>num1:
    print(f"{num2} is large number")
else:
    print(f"Both number are equal")
```

```python
mark = int(input("Enter any number:"))
if mark<35:
    print(f"{mark} is Grade Fail")
elif mark<=49:
    print(f"{mark} is Grade D")
elif mark<=74:
    print(f"{mark} is Grade C")
elif mark<=89:
    print(f"{mark} is Grade B")
else:
    print(f"{mark} is Grade A")
```

```python
num = int(input("Enter any number:"))
if num % 3 == 0 and num % 5 ==0:
    print(f"{num} is FizzBuzz")
elif num % 3 == 0:
    print(f"{num} is Fizz")
elif num % 5 == 0:
    print(f"{num} is Buzz")
else:
    print(f"{num} is  Number itself")
```

```python
character = input("Enter a character:")
if character.isalpha():
    if character.lower() in "aeiou":
        print("Vowel")
    else:
        print("Consonant")
else:
    print("Not an alphabetical character")
```

```python
num1 = int(input("Enter first number:"))
num2 = int(input("Enter second number:"))
operator = input("Enter operator (+,-,*,/):")
if operator == "+":
    print("Result:",num1 + num2)
elif operator == "-":
    print("Result:",num1 - num2)
elif operator == "*":
    print("Result:",num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result",num1/num2)
    else:
        print("Cannot divide by Zero")
else:
    print("Invalid operator")
```

```python
side1 = int(input("Enter side 1:"))
side2 = int(input("Enter side 2:"))
side3 = int(input("Enter side 3:"))
if side1 + side2 <= side3 or side2 + side3 <= side1 or side1 + side3 <= side2:
    print("Not a valid Triangle")
else:
    if side1 == side2 == side3:
        print("Equilateral Triangle")
    elif side1 == side2 or side2 == side3 or side1 == side3:
        print("Isosceles Triangle")
    else:
        print("Scalene Triangle")
```
